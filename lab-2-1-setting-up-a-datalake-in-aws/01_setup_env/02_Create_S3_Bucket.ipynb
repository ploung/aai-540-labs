# Create S3 Bucket
## This is for creating Data Lake (aka database)

```python
import importlib.metadata
import sagemaker
from sagemaker.session import Session

print(importlib.metadata.version("sagemaker"))
print(sagemaker.__file__)
print(Session)
```

```python
import os # Added this - not in lab video
import boto3 # Is a wrapper for Python to interact with AWS
from sagemaker.session import Session

region = (
    boto3.session.Session().region_name
    or os.environ.get("AWS_DEFAULT_REGION")
    or "us-east-1"
)
os.environ["AWS_DEFAULT_REGION"] = region

session = boto3.session.Session(region_name=region)
sagemaker_session = Session(boto_session=session)
bucket = sagemaker_session.default_bucket()

s3 = session.client(
    service_name="s3",
    region_name=region,
)

print("Region:", region)
print("Default bucket:", bucket)
```

```python
setup_s3_bucket_passed = False
```

```python
print("Default bucket: {}".format(bucket))
```

# Verify S3_BUCKET Bucket Creation

```python
from botocore.client import ClientError

response = None

try:
    response = s3.head_bucket(Bucket=bucket)
    print(response)
    setup_s3_bucket_passed = True
except ClientError as e:
    print("[ERROR] Cannot find bucket {} in {} due to {}.".format(bucket, response, e))
```
